**`Check Number Is Positive`**

```python
num=int(input("Enter Number:"))
if num>0:
  print("Number is Positive")
elif num<0:
  print("Number is Negative")
else:
  print("Number Is Zero")
```

**`Check Voting By Age`**

```python
age=int(input("Enter Your Age:"))
if age>=18:
  print("You Are Eligible For Voting")
else:
  x=18-age
  print("You Are Not Eligible For Voting")
  print("These Years  Left You Are Eligible For Voting:",x)
```

**`Check Number Is Even OR Odd`**

```python
number=int(input("Enter Number:"))
if number%2==0:
  print("Number Is Even")
else:
  print("Number Is Odd")
```

**`Total Marks And Grade`**

```python
maths=int(input("Enter Maths Marks:"))
physics=int(input("Enter Physics Marks:"))
chemistry=int(input("Enter Chemistry Marks:"))
english=int(input("Enter English Marks:"))
urdu=int(input("Enter Urdu Marks:"))
total_marks=maths+physics+chemistry+english+urdu
print("Total Marks Of Their Subjects Is:",total_marks)
if total_marks>400:
  print("Grade A")
elif total_marks>300:
  print("Grade B")
elif total_marks>200:
  print("Grade C")
elif total_marks>100:
  print("Grade D")
else:
  print("you are Fail")

```
