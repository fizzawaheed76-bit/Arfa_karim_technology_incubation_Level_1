```python
for i in range(3):
  marks = int(input(f"Enter marks for subject {i+1}:"))
  if marks >= 90:
    print("Grade:","A")
  elif marks >= 80:
    print("Grade:","B")
  elif marks >= 70:
    print("Grade:","C")
  elif marks >= 60:
    print("Grade:","D")
  else:
    print("Grade:","F")
  print("Marks:", marks)
```

```python
import random

target = 89
number_guessing_game = 0
threshold = 6

print("Welcome to the Number Guessing Game!")
print("I'm thinking of a number between 1 and 100.")

while number_guessing_game != target:
    try:
        number_guessing_game = int(input("Enter a number between 1 and 100: "))
    except ValueError:
        print("Invalid input. Please enter a whole number.")
        continue

    if number_guessing_game < 1 or number_guessing_game > 100:
        print("Invalid number. Please guess between 1 and 100.")
    elif abs(number_guessing_game - target) <= threshold and number_guessing_game != target:
        print("You're very close!")
    elif number_guessing_game < target:
        print("Too low! Try again.")
    elif number_guessing_game > target:
        print("Too high! Try again.")

print("You are correct! The number was", target)
print("You win!")
```
