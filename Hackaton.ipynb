```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_csv("/content/drive/MyDrive/patients.csv")
status_counts = df['MARITAL'].value_counts().head(10)
plt.figure(figsize=(10, 6))
plt.bar(status_counts.index.astype(str), status_counts.values)

plt.xlabel("Marital Status")
plt.ylabel("Count")
plt.title("Top 10 Marital Status Distribution")
plt.show()
```
