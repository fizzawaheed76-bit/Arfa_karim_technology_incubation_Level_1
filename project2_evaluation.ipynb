```python
status_count=df['status'].value_counts().reset_index()
status_count.columns=['status','counts']
plt.figure(figsize=(4,4))
plt.plot(status_count['status'],status_count['counts'])
plt.xlabel('status')
plt.ylabel('counts')
plt.title('status')
plt.show()
```

```python
region_count=df['region'].value_counts().reset_index()
region_count.columns=['region','counts']
plt.figure(figsize=(10,6))
plt.pie(region_count['counts'],labels=region_count['region'],autopct='%1.1f%%')
plt.title('region')
plt.show()
```

```python
country_count=df['country_code'].value_counts().reset_index()
country_count.columns=['country_code','counts']
plt.figure(figsize=(10,6))
top_10 = country_count.head(10)
plt.bar(top_10['country_code'],top_10['counts'])
plt.xlabel('country_code')
plt.ylabel('counts')
plt.title('Top 10 Country Code')
plt.tight_layout()
plt.show()
```
